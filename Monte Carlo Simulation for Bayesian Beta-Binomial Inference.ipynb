# Exercise 3: Computing probabilities  

#### Imports and transversal definitions: 

```python
import numpy as np

rng = np.random.default_rng(seed=133)

y_1 = 7
n_1 = 10
y_2 = 80
n_2 = 100
y_3 = 200
n_3 = 300
a = 1
b = 2
capital_phi = 100000
```

#### Defining a generalising function for exercises 3 and 4 to avoid repetition: 

```python
def probability_calc(objective="theta_1", inequality="best"):
    vec_phi = []
    for phi in range(capital_phi):
        # Sample thetas:

        theta_1 = rng.beta(a=(a + y_1), b=(b + n_1 - y_1), size=1)
        theta_2 = rng.beta(a=(a + y_2), b=(b + n_2 - y_2), size=1)
        theta_3 = rng.beta(a=(a + y_3), b=(b + n_3 - y_3), size=1)

        # Calculate deltas:

        delta_1 = theta_1 - theta_2
        delta_2 = theta_2 - theta_3
        delta_3 = theta_3 - theta_1

        # Calculate indicator functions:

        if inequality == "best":
            if objective == "theta_1":
                indicator_1 = delta_1 > 0
                indicator_2 = -delta_3 > 0

            if objective == "theta_2":
                indicator_1 = -delta_1 > 0
                indicator_2 = delta_2 > 0

            if objective == "theta_3":
                indicator_1 = -delta_2 > 0
                indicator_2 = delta_3 > 0

        # Inequality == 'worst':

        if inequality == "worst":
            if objective == "theta_1":
                indicator_1 = delta_1 < 0
                indicator_2 = -delta_3 < 0

            if objective == "theta_2":
                indicator_1 = -delta_1 < 0
                indicator_2 = delta_2 < 0

            if objective == "theta_3":
                indicator_1 = -delta_2 < 0
                indicator_2 = delta_3 < 0

        # Compute the indicator variable for the iteration:

        indicator_phi = indicator_1 * indicator_2

        # Check as good practice:

        if indicator_phi != 1 and indicator_phi != 0:
            raise ValueError(f"indicator computed erroneously at iteration {phi}")
        else:
            vec_phi.append(indicator_phi)

    # Once outside the loop, compute the final probability as the sample average of vec_phi:

    prob = float((sum(vec_phi) / capital_phi)[0])
    print(f"Probability of {objective} being the {inequality} equals: {prob}")
    return prob
```

#### Estimate 3.1 

```python
theta_1_best = probability_calc()
```

#### Estimate 3.2

```python
theta_2_best = probability_calc(objective="theta_2")
```

#### Estimate 3.3

```python
theta_3_best = probability_calc(objective="theta_3")
```

A final check (tolerating some error as the thetas draw from different simulations): 

```python
assert np.isclose((theta_1_best + theta_2_best + theta_3_best), 1, atol=0.005)
print(theta_1_best + theta_2_best + theta_3_best)
```
